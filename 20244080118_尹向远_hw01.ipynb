# 欧几里得算法

```python
""". 实现⼀个名为
gcd的函数，它接受两个整数作为其参数并返回它们的GCD。你可以
假设两个输⼊都是整数，所以你不需要在你的函数中包含任何错误检查。
2. 使⽤你的函数评估以下数字对的GCD：
1. (2018, 2019)
2. (1200, 300)
3. (5040, 60)
3. 如果⼀个或两个参数是负数，你的函数会做什么？这种⾏为有意义吗"""


def gcd(a: int, b: int) -> int:
    for i in range(min(a, b), 0, -1):
        if a % i == 0 and b % i == 0:
            return i
    return 1


print(gcd(2018, 2019))
print(gcd(1200, 300))
print(gcd(5040, 60))
print("如果参数中含有负数，函数会返回1，提前终止而不是进行其他操作")
```

```python
"""在这个问题中，你将得到更多关于条件的练习，并对典型的“编程面试”问题所需的思
维方式有⼀个初步的了解。⼀个正整数  是2的幂，如果  
n =2p
。
对于某个整数 
p ≥0
1. 编写⼀个名为
is_power_of_2
的函数，它接受⼀个正整数作为其唯⼀的参数，并返
回⼀个布尔值，指⽰输⼊是否是2的幂。你可以假设输⼊是⼀个正整数。你不能在
你的解决方案中使用内置的math.sqrt函数。你应该只需要使用除法和模（%）运
算。
b n
2. 将你之前的解决方案推广到⼀个名为
is_power
的函数，它接受两个正整数作为其
参数， 和 ，按那个顺序，并返回⼀个布尔值。
is_power(b,n)
如果  是  的幂
则返回True，否则返回False"""


def is_power_of_2(n: int) -> bool:
    while n > 1 and n % 2 == 0:
        n //= 2
    return n == 1


def is_power(b: int, n: int) -> bool:
    while b > 1 and b % n == 0:
        b //= n
    return n == 1
```
