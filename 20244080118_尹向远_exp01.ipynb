# 实验二：：函数、条件、循环与递归

```python
""" 编写score_level(score)函数：90〜100分返回“优秀”，80〜不足90分返回
“良好”，60〜不足80分返回“及格”，0〜不足60分返回“不及格”。对于超出0〜
100范围的数值，抛出ValueError。"""


def score_level(score):
    if score >= 60 and score < 80:
        return "及格"
    elif score < 90:
        return "良好"
    elif score <= 100:
        return "优秀"
    else:
        return ValueError


"""分别使用for和while计算1到100的整数和，比较两种实现的结果。"""
total_for = 0
for i in range(1, 101):
    total_for += i

total_while = 0
i = 1
while i <= 100:
    total_while += i
    i += 1

print(f"for循环求和结果：{total_for}")
print(f"while循环求和结果：{total_while}")
print(f"两种实现结果相同吗：{total_for == total_while}")

"""编写递归版和迭代版阶乘函数，约定0! = 1，输入为非负整数。比较n = 0、1、5、10时的结果。"""


def factorial_recursive(n):
    if n < 0:
        raise ValueError("输入必须为非负整数")
    if n == 0 or n == 1:
        return 1
    return n * factorial_recursive(n - 1)


def factorial_iterative(n):
    if n < 0:
        raise ValueError("输入必须为非负整数")
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result


for n in [0, 1, 5, 10]:
    print(
        f"n={n}, 递归版结果={factorial_recursive(n)}, 迭代版结果={factorial_iterative(n)}"
    )

"""编写summary(values)函数，接收非空数值列表，以元组返回最小值、最大值和平均值，并通过元组解包接收结果。"""


def summary(values):
    if not values:
        raise ValueError("values不能为空")
    minimum = min(values)
    maximum = max(values)
    average = sum(values) / len(values)
    return (minimum, maximum, average)


values = [12, 4, 8, 15, 6, 10]
min_value, max_value, avg_value = summary(values)
print(f"最小值={min_value}, 最大值={max_value}, 平均值={avg_value}")

"""编写sum_all(*args)函数，返回所有参数之和；分别测试无参数、单个参数和多个参数的情况。"""


def sum_all(*args):
    return sum(args)


print(f"无参数：{sum_all()}")
print(f"单个参数：{sum_all(10)}")
print(f"多个参数：{sum_all(1, 2, 3, 4, 5)}")

"""修正score_level(score)函数：确保分数在0到100之间，超出范围抛出ValueError。"""


def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError("分数必须在0到100之间")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"


print(score_level(95))
print(score_level(85))
print(score_level(70))
print(score_level(40))
```

# 实验三:列表、字典、元组与基础统计

```python
import importlib
import os
import pickle

names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

"""使用zip()构建姓名到成绩的字典，使用items()逐项输出姓名和成绩。"""
score_dict = dict(zip(names, scores))

for name, score in score_dict.items():
    print(f"{name}：{score}")

"""使用enumerate()输出序号、姓名和成绩，序号从1开始。"""
for index, name in enumerate(names, start=1):
    print(f"序号{index}：{name}，成绩{score_dict[name]}")

""" 使用列表推导式提取及格学生的姓名，并生成所有成绩的平方列表。"""
passed_names = [name for name in names if score_dict[name] >= 60]
score_squares = [score**2 for score in scores]
print(f"及格学生姓名：{passed_names}")
print(f"成绩平方列表：{score_squares}")

"""计算人数、平均分、最高分、最低分和及格率，使用字符串格式化将平均分显示为
两位小数、及格率显示为百分数。"""
count = len(scores)
average = sum(scores) / count
pass_rate = len(passed_names) / count

print(f"人数：{count}")
print(f"平均分：{average:.2f}")
print(f"最高分：{max(scores)}")
print(f"最低分：{min(scores)}")
print(f"及格率：{pass_rate:.1%}")

"""比较alias = scores与copied = scores.copy()：修改alias中的一个元素，
观察原列表和副本的变化。该观察任务使用单独的演示列表，避免改变后续实验的
原始成绩。"""
# 使用单独的演示列表，避免改变后续实验使用的原始成绩
demo_scores = [70, 80, 90]
alias = demo_scores
copied = demo_scores.copy()

alias[0] = 0

print(f"演示列表：{demo_scores}")
print(f"直接赋值alias：{alias}")
print(f"copy()副本copied：{copied}")
print(f"原始成绩列表未受影响：{scores}")

""" 使用元组保存一条学生记录，验证其元素不能直接重新赋值；观察⻓度不同的两个
序列使用zip()时的结果。"""
student_record = ("Alice", 85)
print(f"学生记录元组：{student_record}")

try:
    student_record[1] = 90
except TypeError as e:
    print(f"元组元素不能直接重新赋值：{e}")

short_scores = scores[:4]
print(f"names共{len(names)}项，short_scores共{len(short_scores)}项")
print(f"zip结果：{list(zip(names, short_scores))}")

"""6. 将有效记录使用pickle.dump()保存为exp01_data/students.pkl，使用
pickle.load()读取自己生成的文件，验证恢复的数据与原数据相等。"""
valid_records = [{"name": name, "score": score_dict[name]} for name in names]

pickle_file = os.path.join("exp01_data", "students.pkl")
os.makedirs("exp01_data", exist_ok=True)

with open(pickle_file, "wb") as f:
    pickle.dump(valid_records, f)

with open(pickle_file, "rb") as f:
    loaded_records = pickle.load(f)

print(f"有效记录：{valid_records}")
print(f"恢复的记录：{loaded_records}")
print(f"恢复数据与原数据相等：{loaded_records == valid_records}")

"""7. 将parse_student()保存到Notebook同级的student_utils.py模块中，通过
import调用，并展示一个有效输入和一个无效输入的测试结果。"""
module_source = '''
def parse_student(line):
    """将"姓名,成绩"格式的单行记录解析为字典，无效记录抛出ValueError。"""
    fields = line.strip().split(",")
    if len(fields) != 2:
        raise ValueError(f"字段数量应为2，实际为{len(fields)}")

    name = fields[0].strip()
    if not name:
        raise ValueError("姓名不能为空")

    try:
        score = int(fields[1].strip())
    except ValueError:
        raise ValueError(f"成绩必须是整数，实际为{fields[1].strip()}") from None

    if not 0 <= score <= 100:
        raise ValueError(f"成绩必须在0~100之间，实际为{score}")

    return {"name": name, "score": score}
'''

module_path = "student_utils.py"
with open(module_path, "w", encoding="utf-8") as f:
    f.write(module_source)

import student_utils

importlib.reload(student_utils)
parse_student = student_utils.parse_student

valid_line = "Alice,85"
print(f"有效输入 {valid_line} → {parse_student(valid_line)}")

try:
    parse_student("Bob,abc")
except ValueError as e:
    print(f"无效输入 Bob,abc → {e}")

"""8. 读取一个明确不存在的文件，捕获FileNotFoundError并输出说明"""
missing_path = os.path.join("exp01_data", "not_exist.txt")
try:
    with open(missing_path, encoding="utf-8") as f:
        f.read()
except FileNotFoundError as e:
    print(f"打开不存在的文件出错：{e}")
    print(
        "说明：路径不存在时open()会抛出FileNotFoundError，可用try/except捕获并给出提示"
    )
```

# 实验4：文件读写、异常处理与模块

```python
import os

from student_utils import parse_student

"""1.使用os.getcwd()查看工作目录，使用os.path.join()构造相对路径，并使用
os.makedirs(..., exist_ok=True)创建数据目录。"""
print(f"当前工作目录：{os.getcwd()}")

DATA_DIR = "exp01_data"
DATA_FILE = os.path.join(DATA_DIR, "students.txt")
REPORT_FILE = os.path.join(DATA_DIR, "report.txt")

os.makedirs(DATA_DIR, exist_ok=True)
print(f"数据目录已存在或创建成功：{DATA_DIR}")
print(f"数据文件相对路径：{DATA_FILE}")

"""2. 使用with open(..., encoding="utf-8")写入上述数据，再读取文件，跳过表
头，使用字符串方法解析记录。"""
lines = ["name,score"]
lines.extend(f"{name},{score}" for name, score in zip(names, scores))
lines.extend(["Grace,abc", "Henry,105", "Incomplete"])

with open(DATA_FILE, "w", encoding="utf-8") as f:
    f.write("\n".join(lines) + "\n")

print("写入的文件内容：")
with open(DATA_FILE, encoding="utf-8") as f:
    print(f.read())

print("跳过表头后，使用字符串方法解析：")
with open(DATA_FILE, encoding="utf-8") as f:
    header = f.readline().strip()
    print(f"表头：{header}")
    for line in f:
        line = line.strip()
        if not line:
            continue
        name, _, score_text = line.partition(",")
        print(f"姓名={name.strip()!r}, 成绩原文={score_text.strip()!r}")

"""3. 编写parse_student(line)函数：将成绩转换为整数，检查字段数量、姓名是否
为空及成绩是否在0〜100范围内。无效记录抛出ValueError。"""
print(f"有效记录：{parse_student('Alice,85')}")

for bad_line in ["Grace,abc", "Henry,105", "Incomplete"]:
    try:
        parse_student(bad_line)
    except ValueError as e:
        print(f"无效记录 {bad_line!r}：{e}")

"""4. 按行处理文件，捕获无效记录的异常，记录行号与错误原因，并继续处理后续记
录。有效记录保存为字典组成的列表。"""
valid_records = []
invalid_records = []

with open(DATA_FILE, encoding="utf-8") as f:
    f.readline()  # 跳过表头
    for line_number, line in enumerate(f, start=2):
        try:
            valid_records.append(parse_student(line))
        except ValueError as e:
            invalid_records.append((line_number, str(e)))
            print(f"第{line_number}行无效：{e}")

print(f"有效记录共{len(valid_records)}条：{valid_records}")
print(f"无效记录共{len(invalid_records)}条：{invalid_records}")

"""5. 将有效记录数、无效记录数、平均分和及格率写入exp01_data/report.txt。分
别演示w与a模式，说明二者对已有内容的影响。"""
valid_count = len(valid_records)
score_values = [record["score"] for record in valid_records]
average_score = sum(score_values) / valid_count
pass_rate = sum(score >= 60 for score in score_values) / valid_count

with open(REPORT_FILE, "w", encoding="utf-8") as f:
    f.write(f"有效记录数：{valid_count}\n")
    f.write(f"无效记录数：{len(invalid_records)}\n")
    f.write(f"平均分：{average_score:.2f}\n")
    f.write(f"及格率：{pass_rate:.1%}\n")

with open(REPORT_FILE, encoding="utf-8") as f:
    print("w模式写入后读取：")
    print(f.read())

with open(REPORT_FILE, "a", encoding="utf-8") as f:
    f.write("追加模式测试行\n")

with open(REPORT_FILE, encoding="utf-8") as f:
    print("a模式追加后读取：")
    print(f.read())

print("说明：w模式会先清空文件再写入，a模式保留原内容并在末尾追加。")
```

# 实验5：类、继承与对象复制

```python
import copy

"""1. 定义Student类，在__init__()中初始化name和score属性，并实现is_passed()
和__str__()方法。"""


class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"


"""2. 为实验4中的每条有效记录创建一个Student对象，输出对象信息和及格状态。"""
students = [Student(record["name"], record["score"]) for record in valid_records]

for student in students:
    status = "及格" if student.is_passed() else "不及格"
    print(f"{student}，及格状态：{status}")

"""3. 定义继承自Student的GraduateStudent类，增加research_topic属性，通过
super().__init__()初始化父类属性，并重写__str__()。"""


class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return (
            f"GraduateStudent(name={self.name}, score={self.score}, "
            f"research_topic={self.research_topic})"
        )


graduate_student = GraduateStudent("Alice", 85, "机器学习")
print(graduate_student)
print(f"及格状态：{graduate_student.is_passed()}")

""""4. 构造包含嵌套列表的对象，例如{"name": "Alice", "scores": [85, 90]}。分别进行直接赋值、copy.copy()和copy.deepcopy()，修改内部列表，比较结果。
每次比较都从新建的原始对象开始。"""


def build_original():
    return {"name": "Alice", "scores": [85, 90]}


original = build_original()
alias = original
alias["scores"].append(100)
print("直接赋值（alias = original）：")
print(f"  original = {original}")
print(f"  alias    = {alias}")
print(f"  两者是同一个对象：{alias is original}")

original = build_original()
shallow = copy.copy(original)
shallow["scores"].append(100)
print("浅拷贝（copy.copy）：")
print(f"  original = {original}")
print(f"  shallow  = {shallow}")
print(f"  内部列表是同一个对象：{shallow['scores'] is original['scores']}")

original = build_original()
deep = copy.deepcopy(original)
deep["scores"].append(100)
print("深拷贝（copy.deepcopy）：")
print(f"  original = {original}")
print(f"  deep     = {deep}")
print(f"  内部列表是同一个对象：{deep['scores'] is original['scores']}")
```
