## 字符串的乐趣

```python
def is_palindrome(text):
    """忽略空格和大小写，判断字符串是否为回文。"""
    compact = text.replace(" ", "").lower()
    return compact == compact[::-1]


def is_abecedarian(text):
    """忽略空格，判断字母是否按非降序排列。"""
    letters = text.replace(" ", "").lower()
    return all(left <= right for left, right in zip(letters, letters[1:]))


def remove_vowels(text):
    """删除字符串中的所有英文元音，并保留其余字符原有大小写。"""
    vowels = "aeiouAEIOU"
    return "".join(character for character in text if character not in vowels)


palindrome_tests = ["level", "T A C O cat", "rats live on no evil star", "abc"]
for value in palindrome_tests:
    print(f"is_palindrome({value!r}) -> {is_palindrome(value)}")

for value in ["adder", "abcd efgh xyz", "dog", "beet"]:
    print(f"is_abecedarian({value!r}) -> {is_abecedarian(value)}")

for value in ["cat", "audio", "AEIOU", "rhythm"]:
    print(f"remove_vowels({value!r}) -> {remove_vowels(value)!r}")

assert is_palindrome("T A C O cat")
assert not is_palindrome("python")
assert is_abecedarian("abcd efgh xyz")
assert not is_abecedarian("dog")
assert remove_vowels("cat") == "ct"
assert remove_vowels("audio") == "d"
assert remove_vowels("AEIOU") == ""
```

## 列表的乐趣

```python
def list_reverse(items):
    """返回列表的反转副本，并检查输入类型。"""
    if not isinstance(items, list):
        raise TypeError("list_reverse 的参数必须是 list")
    return items[::-1]


def is_sorted(seq):
    """仅遍历序列一次，判断元素是否按升序排列。"""
    iterator = iter(seq)
    try:
        previous = next(iterator)
    except StopIteration:
        return True

    for current in iterator:
        if current < previous:
            return False
        previous = current
    return True


print(f"list_reverse([1, 2, 3]) -> {list_reverse([1, 2, 3])}")
print(f"is_sorted([1, 2, 2, 5]) -> {is_sorted([1, 2, 2, 5])}")
print(f"is_sorted((1, 3, 2)) -> {is_sorted((1, 3, 2))}")
print(f"is_sorted('abcd') -> {is_sorted('abcd')}")

assert list_reverse([1, 2, 3]) == [3, 2, 1]
assert is_sorted([1, 2, 2, 5])
assert not is_sorted([2, 1])
try:
    list_reverse((1, 2, 3))
except TypeError as error:
    print(f"预期异常：{error}")
else:
    raise AssertionError("非列表输入应触发 TypeError")
```

## 作为向量的元组

```python
def _is_number(value):
    return isinstance(value, (int, float)) and not isinstance(value, bool)


def vec_scalar_mult(vector, scalar):
    """返回元组向量与数值标量的乘积。"""
    if not isinstance(vector, tuple):
        raise TypeError("vector 必须是元组")
    if not all(_is_number(value) for value in vector):
        raise TypeError("vector 的元素必须是整数或浮点数")
    if not _is_number(scalar):
        raise TypeError("scalar 必须是整数或浮点数")
    return tuple(value * scalar for value in vector)


def vec_inner_product(left, right):
    """返回两个元组向量的内积。"""
    if not isinstance(left, tuple) or not isinstance(right, tuple):
        raise TypeError("两个输入都必须是元组")
    if not all(_is_number(value) for value in left + right):
        raise TypeError("向量元素必须是整数或浮点数")
    if len(left) != len(right):
        raise ValueError("两个向量的长度必须一致")
    return float(sum(x * y for x, y in zip(left, right)))


print(f"vec_scalar_mult((1, 2, 3), 2.5) -> {vec_scalar_mult((1, 2, 3), 2.5)}")
print(f"vec_inner_product((1, 2), (3, 4)) -> {vec_inner_product((1, 2), (3, 4))}")

assert vec_scalar_mult((1, 2), 3) == (3, 6)
assert vec_inner_product((1, 2), (3, 4)) == 11.0
```

```python
def check_valid_mx(matrix):
    """检查参数是否为元素和长度均合法的元组矩阵。"""
    if not isinstance(matrix, tuple):
        return False
    if not all(isinstance(row, tuple) for row in matrix):
        return False
    if not all(_is_number(value) for row in matrix for value in row):
        return False
    row_lengths = {len(row) for row in matrix}
    return len(row_lengths) <= 1


def mx_vec_mult(matrix, vector):
    """计算矩阵与列向量的乘积，返回数值元组。"""
    if not check_valid_mx(matrix):
        raise ValueError("matrix 必须是有效的元组矩阵")
    if not isinstance(vector, tuple):
        raise TypeError("vector 必须是元组")
    if not all(_is_number(value) for value in vector):
        raise TypeError("vector 的元素必须是整数或浮点数")
    if matrix and len(matrix[0]) != len(vector):
        raise ValueError("矩阵列数必须等于向量维数")
    return tuple(sum(value * weight for value, weight in zip(row, vector)) for row in matrix)


sample_matrix = ((1, 2), (3, 4), (5, 6))
sample_vector = (2, 3)
print(f"check_valid_mx(sample_matrix) -> {check_valid_mx(sample_matrix)}")
print(f"mx_vec_mult({sample_matrix}, {sample_vector}) -> {mx_vec_mult(sample_matrix, sample_vector)}")

assert check_valid_mx(sample_matrix)
assert not check_valid_mx(((1, 2), (3,)))
assert mx_vec_mult(sample_matrix, sample_vector) == (8, 18, 28)
```

## 更多向量的乐趣

```python
class Vector:
    def __init__(self, dimension, elements=None):
        if not isinstance(dimension, int) or isinstance(dimension, bool):
            raise TypeError("dimension 必须是非负整数")
        if dimension < 0:
            raise ValueError("dimension 不能为负数")
        if elements is None:
            elements = (0,) * dimension
        if not isinstance(elements, (list, tuple)):
            raise TypeError("elements 必须是列表或元组")
        if len(elements) != dimension:
            raise ValueError("elements 的个数必须等于 dimension")
        if not all(_is_number(value) for value in elements):
            raise TypeError("elements 只能包含整数或浮点数")

        self.dimension = dimension
        self.elements = tuple(elements)

    def _require_vector(self, other):
        if not isinstance(other, Vector):
            raise TypeError("比较对象必须是 Vector")
        if self.dimension != other.dimension:
            raise ValueError("两个向量的维度必须相同")

    def __repr__(self):
        return f"Vector({self.dimension}, {self.elements})"

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        self._require_vector(other)
        return self.elements == other.elements

    def __ne__(self, other):
        result = self.__eq__(other)
        return NotImplemented if result is NotImplemented else not result

    def __lt__(self, other):
        self._require_vector(other)
        return self.elements < other.elements

    def __le__(self, other):
        self._require_vector(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        self._require_vector(other)
        return self.elements > other.elements

    def __ge__(self, other):
        self._require_vector(other)
        return self.elements >= other.elements

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("dot 的参数必须是 Vector")
        if self.dimension != other.dimension:
            raise ValueError("两个向量的维度必须相同")
        return float(sum(x * y for x, y in zip(self.elements, other.elements)))

    def __mul__(self, other):
        if isinstance(other, Vector):
            if self.dimension != other.dimension:
                raise ValueError("两个向量的维度必须相同")
            return Vector(
                self.dimension,
                tuple(x * y for x, y in zip(self.elements, other.elements)),
            )
        if _is_number(other):
            return Vector(self.dimension, tuple(x * other for x in self.elements))
        return NotImplemented

    def __rmul__(self, other):
        return self.__mul__(other)


zero_vector = Vector(4)
left = Vector(2, (2, 4))
right = Vector(2, (2, 5))
product_left = Vector(2, (2, 3))
product_right = Vector(2, (4, 5))

print(f"默认零向量：{zero_vector}")
print(f"{left} < {right}：{left < right}")
print(f"{left} == {Vector(2, (2, 4))}：{left == Vector(2, (2, 4))}")
print(f"2 * {product_left} = {2 * product_left}")
print(f"{product_left} * 2 = {product_left * 2}")
print(f"{product_left} * {product_right} = {product_left * product_right}")
print(f"{product_left}.dot({product_right}) = {product_left.dot(product_right)}")

assert Vector(4) == Vector(4, (0, 0, 0, 0))
assert left < right
assert product_left * product_right == Vector(2, (8, 15))
assert 2 * product_left == Vector(2, (4, 6))
assert 2 * product_left == product_left * 2
assert product_left.dot(product_right) == 23.0
```

```python
# 补充异常测试
error_cases = [
    lambda: Vector(2, (1,)),
    lambda: Vector(-1),
    lambda: Vector(2) == Vector(3),
    lambda: Vector(2, (1, 2)).dot(Vector(3, (1, 2, 3))),
    lambda: Vector(2, (1, 2)) * Vector(3, (1, 2, 3)),
]

for case_number, case in enumerate(error_cases, start=1):
    try:
        case()
    except (TypeError, ValueError) as error:
        print(f"异常测试 {case_number}：{type(error).__name__}: {error}")
    else:
        raise AssertionError(f"异常测试 {case_number} 未触发异常")
```
